```python
# convert to numpy array
feauter_cols = iris.feature_names
x = df[feauter_cols]
y = df["target"]
```

```python
x = x.to_numpy()
y = y.to_numpy()
```

```python
# Numpy stats (coloumn-wise): mean., median, std, var, min, max
coloum_mean = np.mean(x, axis=0)
coloum_median = np.median(x, axis=0)
coloum_std = np.std(x, axis=0)
coloum_var = np.var(x, axis=0)
coloum_min = np.min(x, axis=0)
coloum_max = np.max(x, axis=0)

print("Column-wise mean:", coloum_mean)
print("Column-wise median:", coloum_median)
print("Column-wise standard deviation:", coloum_std)
print("Column-wise variance:", coloum_var) 
print("Column-wise minimum:", coloum_min)
print("Column-wise maximum:", coloum_max)   
```

```python
print("Missing values in x:", np.isnan(x).sum())
print("Missing values in y:", df.isnull().sum())
```

```python
df.groupby("Species").size()
```

```python
df.groupby("Species")[feauter_cols].mean()
```

```python
# pair plot with seaborn
sns.set_style("whitegrid")  

sns.pairplot(df[feauter_cols + ["Species"]], hue="Species", palette="Set2", diag_kind="kde", markers=["o", "s", "D"]  )
```

```python
# correlation heat MAP 

corr = df[feauter_cols].corr(numeric_only=True)
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlation Heatmap of Iris Features")
```

```python
# class distribution plot

plt.figure(figsize=(6, 4))
sns.countplot(x="Species", data=df, palette="Set2")
plt.title("Class Distribution of Iris Species")
plt.show()
```

```python
# scatter plot of sepal length vs sepal width
plt.figure(figsize=(8, 6))
sns.scatterplot(x="sepal length (cm)", y="sepal width (cm)", hue="Species", data=df, palette="Set2", s=100)
plt.title("Scatter Plot of Sepal Length vs Sepal Width")
plt.show()
```
