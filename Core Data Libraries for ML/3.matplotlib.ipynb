**❓ What is Matplotlib**

Matplotlib is a Python library for creating visual charts and graphs. It allows you to plot data using lines, bars, histograms, scatter plots, and many other styles.

**⚙️ Why Matplotlib is Important in Machine Learning**

Matplotlib is the **foundation of data visualization** in Python.  
It helps us understand data patterns, relationships, and model performance through visual insights.

**🔍 Why Matplotlib is Essential:**
- 📊 **Data Exploration:** Visualize data distributions, correlations, and trends.
- 🤖 **Model Understanding:** Plot decision boundaries, training curves, feature importance.
- 📈 **Performance Evaluation:** Compare predictions vs actuals, loss curves, accuracy.
- 🧩 **Integration:** Works perfectly with NumPy and Pandas.

> 💡 *If NumPy handles numbers and Pandas handles data, Matplotlib helps you SEE them.*

### Simple Line plot

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

print("x:", x[:5])
print("y:", y[:5])


plt.plot(x, y)
plt.title("Sine Wave")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.show()
```

### Customized plot

```python
plt.figure(figsize=(15, 4))
plt.plot(x, y, label="sin(x)", color="blue", marker="o", linestyle="--")
plt.title("Sine Wave")
plt.xlabel("x Values")
plt.ylabel("sin(x)")
plt.legend()
plt.grid()
plt.show()
```

### Bar Chart

```python
students = ['Alice', 'Bob', 'Charlie', 'David', 'Eva']
scores = [85, 90, 78, 92, 88]

plt.bar(students, scores, color='skyblue')
plt.title("Student Scores - Bar Chart")
plt.xlabel("Students")
plt.ylabel("Scores")
plt.show()  
```

```python
plt.barh(students, scores, color='orange')
plt.title("Student Scores -  horizontal Bar Chart")
plt.ylabel("Students")
plt.xlabel("Scores")
plt.show()  
```

### Scatter Plot

```python
np.random.seed(42)
x = np.random.rand(50)
y = np.random.rand(50)

print(x[:5])
print(y[:5])

plt.scatter(x, y, color='green', marker='o', alpha=0.7)
plt.title("Random Scatter Plot")
plt.xlabel("feauter x")
plt.ylabel("feature y")
plt.show()
```

### Histogram  - its manily used for distribution , this is an essentils for checking skewness of Data . how normalised or skeweed

```python
data = np.random.rand(1000)

plt.hist(data, bins=30, color='purple', edgecolor='black')
plt.title("Random Data Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

### Pie - Chart

```python
labels = ["java", "python", "c++", "javascript"]
sizes = [30, 40, 20, 10]

plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=140)
plt.title("Programming Language Popularity")
plt.axis('equal')
plt.show()
```

### Mutliple Plotes (Sub-plots)

```python
x = np.linspace(0, 10, 100)
plt.Figure(figsize=(15, 4))

plt.subplot(1,2,1)
plt.plot(x, np.sin(x), "r--")
plt.title("Sine Wave")

plt.subplot(1,2,2)
plt.plot(x, np.cos(x), "b--") 
plt.title("Cosine Wave")  



plt.suptitle("Sine and Cosine Waves")
plt.show()
```
